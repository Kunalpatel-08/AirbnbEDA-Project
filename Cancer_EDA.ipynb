# Discriptive Analysis

```python
# Age column analysis (numeric dtype)
age = data["Age"].dropna()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.kdeplot(
    x=age,
    fill=True,
    color="#4682B4",
    linewidth=2,
    ax=axes[0]
)
axes[0].set_title("Age Distribution (KDE)")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Density")

sns.histplot(
    x=age,
    bins=30,
    kde=True,          # overlays KDE on the histogram
    color="#FF7F50",
    edgecolor="black",
    alpha=0.7,
    ax=axes[1]
)
axes[1].set_title("Age Distribution (Histogram)")
axes[1].set_xlabel("Age")
axes[1].set_ylabel("Count")

fig.tight_layout()
plt.show()
```

```python
# Gender column analysis ( Categorical dtypr)
# Calculate counts once and exclude missing values
gender_counts = data["Gender"].dropna().value_counts()
print(gender_counts)
```

```python
# visualise barplot for Gender distribution analysis

fig, ax = plt.subplots(figsize=(7, 4))

sns.barplot(
    x=gender_counts.index,
    y=gender_counts.values,
    hue=gender_counts.index,
    palette=["#4C78A8", "#F585B3", "#54A24B"],
    legend=False,
    ax=ax
)

# Add count labels above bars
for bar, count in zip(ax.patches, gender_counts.values):
    ax.annotate(
        f"{count:,}",
        (bar.get_x() + bar.get_width() / 2, bar.get_height()),
        ha="center",
        va="bottom",
        xytext=(0, 4),
        textcoords="offset points"
    )

ax.set_title("Gender Distribution", fontsize=14, fontweight="bold")
ax.set_xlabel("Gender")
ax.set_ylabel("Count")
ax.grid(axis="y", linestyle="--", alpha=0.4)

sns.despine()
fig.tight_layout()
plt.show()
```

### Insights

The dataset contains three gender categories(Male, Female ,other) with the most common being Male (16,796 records).
Gender distribution is sufficient for evaluating gender-specific survival trends and severity outcomes.
