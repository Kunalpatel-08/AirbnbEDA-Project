```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("zahidmughal2343/global-cancer-patients-2015-2024")

print("Path to dataset files:", path)
```

```python
import os

print(os.listdir(path))
```

```python


csv_file = next(file for file in os.listdir(path) if file.endswith(".csv"))

data = pd.read_csv(os.path.join(path,csv_file))
```

# Discriptive Analysis

```python
# Age column analysis (numeric dtype)
age = data["Age"].dropna()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.kdeplot(
    x=age,
    fill=True,
    color="#4682B4",
    linewidth=2,
    ax=axes[0]
)
axes[0].set_title("Age Distribution (KDE)")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Density")

sns.histplot(
    x=age,
    bins=30,
    kde=True,          # overlays KDE on the histogram
    color="#FF7F50",
    edgecolor="black",
    alpha=0.7,
    ax=axes[1]
)
axes[1].set_title("Age Distribution (Histogram)")
axes[1].set_xlabel("Age")
axes[1].set_ylabel("Count")

fig.tight_layout()
plt.show()
```

```python
print(data["Age"].describe())
```

### Insights

Range: 20 to 89 years

Mean Age: 54.42 years

Standard Deviation: 20.22

Interquartile Range (IQR): 37 (Q1) to 72 (Q3)

This suggests a broad representation of both young and elderly patients in the dataset, which supports age-based comparative analysis.

```python
# Gender column analysis ( Categorical dtypr)
# Calculate counts once and exclude missing values
gender_counts = data["Gender"].dropna().value_counts()
print(gender_counts)
```

```python
# visualise barplot for Gender distribution analysis

fig, ax = plt.subplots(figsize=(7, 4))

sns.barplot(
    x=gender_counts.index,
    y=gender_counts.values,
    hue=gender_counts.index,
    palette=["#4C78A8", "#F585B3", "#54A24B"],
    legend=False,
    ax=ax
)

# Add count labels above bars
for bar, count in zip(ax.patches, gender_counts.values):
    ax.annotate(
        f"{count:,}",
        (bar.get_x() + bar.get_width() / 2, bar.get_height()),
        ha="center",
        va="bottom",
        xytext=(0, 4),
        textcoords="offset points"
    )

ax.set_title("Gender Distribution", fontsize=14, fontweight="bold")
ax.set_xlabel("Gender")
ax.set_ylabel("Count")
ax.grid(axis="y", linestyle="--", alpha=0.4)

sns.despine()
fig.tight_layout()
plt.show()
```

### Insights

The dataset contains three gender categories(Male, Female ,other) with the most common being Male (16,796 records).
Gender distribution is sufficient for evaluating gender-specific survival trends and severity outcomes.

```python
# Country cloumn analysis

country_counts = data["Country_Region"].dropna().value_counts()
print(country_counts)
```

```python
# Visualise piechart for Country wise distribution analysis

colors = plt.cm.Set3.colors[:len(country_counts)]

plt.figure(figsize=(5,5))
plt.pie(
    country_counts.values,
    labels=country_counts.index,
    colors = colors,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"edgecolor": "white", "linewidth": 1.2},

    textprops={"fontsize":9}
)

plt.title("COuntry/Region Distributor", fontsize=15, fontweight="bold")
plt.axis("equal")
plt.tight_layout()
plt.show()
```

### Insights


*   Patients come from 10 different countries/regions, with Australia being the most represented (5,092 patients i.e about 10.2%).

*   Number of data points for each country is almost same.

*   This diversity enables cross-country comparison of cancer outcomes and treatment economic.

```python
# Cancer Type analysis

cancer_type_counts = data["Cancer_Type"].dropna().value_counts().sort_values(ascending=False)

cancer_percentages = (cancer_counts / cancer_counts.sum() * 100).sort_values(ascending=False)

print(cancer_type_counts)
print(f"---------------------------------------------")
print(cancer_percentages)
```

```python
# Visualise Horizontal barchart  for Cancer Type wise distribution analysis


plt.figure(figsize=(8, 4))

ax = sns.barplot(
    x=cancer_type_counts.values,
    y=cancer_type_counts.index,
    hue=cancer_type_counts.index,
    palette="viridis",
    legend=False
)

for i, (count, percentage) in enumerate(zip(cancer_type_counts.values, cancer_percentages.values)):
    ax.text(
        count,
        i,
        f"  {count} ({percentage:.1f}%)",
        va="center",
        fontsize=10
    )

plt.title("Cancer Type Count and Percentage", fontsize=14, fontweight="bold", pad=8)
plt.xlabel("Count", fontsize=10)
plt.ylabel("Cancer Type", fontsize=10)
plt.grid(axis="x", linestyle="--", alpha=0.35)
sns.despine(top=True, right=True)

plt.tight_layout()
plt.show()
```

### Insights :

We have in  total 8 types of cancer , with each cancer having approx same number of data points under the label , most common cancer are ***Colon cancer ( 6376 patients i.e about 12.8%)*** followed by ***Prostate cancer ( 6308 patients i.e about 12.6%)***
